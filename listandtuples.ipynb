```python
#4
P_lang=["C","C++","java","Python"]
print("The initial list of languages is : ",P_lang)
P_lang[1]="C#"

print("\n The list after replacing second language with the new one is : ",P_lang)
P_lang.append("SQL")
print("\n The language added at the end using append is : ",P_lang)
P_lang.insert(1,"PHP")
print("\n The language added using insert at second position is : ",P_lang)

```

```python
#10
marks=(78,89,56,67,45)
print("the total number of marks are : ",len(marks))
highest=max(marks)
print("\n The highest marks are : ",highest)
lowest=min(marks)
print("\n The lowest marks are : ",lowest)
average=sum(marks)/len(marks)
print("\n The average marks are : ",average)

```
